```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
```

```python

# Load the FIFA 21 dataset and disable low_memory warnings
df = pd.read_csv("C:/Users/hp/Downloads/DataCleaning/Data/fifa21_raw_data.csv", low_memory=False)

# Display the first five rows of the dataset
df.head()
```

```python
# Display the Height and Weight columns
df[["Height", "Weight"]]

# Check the datatypes of Height and Weight columns
df[["Height", "Weight"]].dtypes

# OR (more detailed info)
df[["Height", "Weight"]].info()
```

```python
# Split feet and inches easily
df["Feet"] = df["Height"].str.split("'").str[0].astype(float)
df["Inch"] = df["Height"].str.split("'").str[1].str.replace('"', '').astype(float)
# Convert to cm
df["Height"] = df["Feet"] * 30.48 + df["Inch"] * 2.54
#convert "lbs" to "kg"
df["Weight"] = df["Weight"].str.replace("lbs", "").astype(float) * 0.453592
```

```python
# Convert Joined column into datetime format
df["Joined"] = pd.to_datetime(df["Joined"])

# Split into year, month, day
df["Joined_Year"] = df["Joined"].dt.year
df["Joined_Month"] = df["Joined"].dt.month
df["Joined_Day"] = df["Joined"].dt.day

# Show results
df[["Joined", "Joined_Year", "Joined_Month", "Joined_Day"]].head()
```

```python
df[["Value","Wage","Release Clause"]]
```

```python
import numpy as np

# Remove € sign
df["Value"] = df["Value"].str.replace("€", "", regex=False)
df["Wage"] = df["Wage"].str.replace("€", "", regex=False)
df["Release Clause"] = df["Release Clause"].str.replace("€", "", regex=False)

# Replace missing or non-numeric entries with NaN
df["Value"] = df["Value"].replace(["-", "--", "Free", ""], np.nan)
df["Wage"] = df["Wage"].replace(["-", "--", "Free", ""], np.nan)
df["Release Clause"] = df["Release Clause"].replace(["-", "--", "Free", ""], np.nan)

# --- VALUE ---
df.loc[df["Value"].str.contains("M", na=False), "Value"] = (
    df.loc[df["Value"].str.contains("M", na=False), "Value"]
      .str.replace("M", "", regex=False).astype(float) * 1_000_000
)

df.loc[df["Value"].str.contains("K", na=False), "Value"] = (
    df.loc[df["Value"].str.contains("K", na=False), "Value"]
      .str.replace("K", "", regex=False).astype(float) * 1_000
)

# --- WAGE ---
df.loc[df["Wage"].str.contains("M", na=False), "Wage"] = (
    df.loc[df["Wage"].str.contains("M", na=False), "Wage"]
      .str.replace("M", "", regex=False).astype(float) * 1_000_000
)

df.loc[df["Wage"].str.contains("K", na=False), "Wage"] = (
    df.loc[df["Wage"].str.contains("K", na=False), "Wage"]
      .str.replace("K", "", regex=False).astype(float) * 1_000
)

# --- RELEASE CLAUSE ---
df.loc[df["Release Clause"].str.contains("M", na=False), "Release Clause"] = (
    df.loc[df["Release Clause"].str.contains("M", na=False), "Release Clause"]
      .str.replace("M", "", regex=False).astype(float) * 1_000_000
)

df.loc[df["Release Clause"].str.contains("K", na=False), "Release Clause"] = (
    df.loc[df["Release Clause"].str.contains("K", na=False), "Release Clause"]
      .str.replace("K", "", regex=False).astype(float) * 1_000
)

# Final convert to integers
df["Value"] = df["Value"].astype(float).astype(int)
df["Wage"] = df["Wage"].astype(float).astype(int)
df["Release Clause"] = df["Release Clause"].astype(float).astype(int)
```

```python
# Remove newline characters (\n) and extra spaces from the Hits column
df["Hits"] = df["Hits"].str.replace("\n", "", regex=False).str.strip()
# Display cleaned Hits column
df["Hits"].head()
```

```python
# Step 1: Remove newline characters and extra spaces
df["Team & Contract"] = df["Team & Contract"].str.replace("\n", " ", regex=False).str.strip()

# Step 2: Replace multiple spaces with a single space
df["Team & Contract"] = df["Team & Contract"].replace(r"\s+", " ", regex=True)

# Step 3: Split into Team (text part) and Contract (year range)
# We split on the first occurrence of a digit (start of "2004")
df["Team"] = df["Team & Contract"].str.extract(r"([A-Za-z\s\.]+)")
df["Contract"] = df["Team & Contract"].str.extract(r"(\d{4}.*)")

# Step 4: Clean Team & Contract fields
df["Team"] = df["Team"].str.strip()
df["Contract"] = df["Contract"].str.strip()
```
